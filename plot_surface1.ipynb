```python
import numpy as np
import matplotlib.pyplot as plt


# Define activation functions
def sigmoid(z):
    """Sigmoid activation function"""
    return 1 / (1 + np.exp(-z))

def hard_limit(z):
    """Hard limit activation (step function)"""
    return np.where(z >= 0, 1, 0)

def radial_basis(z):
    """Radial Basis Function (Gaussian)"""
    return np.exp(-(z**2))

# Define perceptron linear combination
def perceptron_output(x1, x2, activation_func):
    z = -4.79 * x1 + 5.90 * x2 - 0.93
    return activation_func(z)


# Plot surface using a meshgrid (structured grid points)
def plot_surface(n_points, activation_func, title):
    # Generate grid of x1, x2 values in [-2, 2]
    x1 = np.linspace(-2, 2, int(np.sqrt(n_points)))
    x2 = np.linspace(-2, 2, int(np.sqrt(n_points)))
    X1, X2 = np.meshgrid(x1, x2)
    
    # Compute perceptron output
    Y = perceptron_output(X1, X2, activation_func)
    
    # Plot 3D surface
    fig = plt.figure(figsize=(6, 5))
    ax = fig.add_subplot(111, projection='3d')
    surf = ax.plot_surface(X1, X2, Y, cmap="cividis")
    
    # Axis labels and title
    ax.set_xlabel("x1")
    ax.set_ylabel("x2")
    ax.set_zlabel("y")
    ax.set_title(title)

    # Make pane backgrounds white
    ax.xaxis.set_pane_color((1, 1, 1, 1))
    ax.yaxis.set_pane_color((1, 1, 1, 1))
    ax.zaxis.set_pane_color((1, 1, 1, 1))

    # Make grid lines very light
    ax.xaxis._axinfo["grid"]['color'] = (0.9, 0.9, 0.9, 1)
    ax.yaxis._axinfo["grid"]['color'] = (0.9, 0.9, 0.9, 1)
    ax.zaxis._axinfo["grid"]['color'] = (0.9, 0.9, 0.9, 1)

    plt.show()


# Case (a) 100 structured grid points
plot_surface(100, sigmoid, "Sigmoid Activation (100 sample points)")
plot_surface(100, hard_limit, "Hard Limit Activation (100 sample points)")
plot_surface(100, radial_basis, "RBF Activation (100 sample points)")

```

```python
plot_surface(5000, sigmoid, "Sigmoid Activation (5000 sample points)")
plot_surface(5000, hard_limit, "Hard Limit Activation (5000 sample points)")
plot_surface(5000, radial_basis, "RBF Activation (5000 sample points)")
```
