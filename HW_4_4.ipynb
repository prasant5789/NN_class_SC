```python
import numpy as np
import matplotlib.pyplot as plt

plt.rcParams.update({
    'font.size': 12,
    'axes.titlesize': 12,
    'axes.labelsize': 12,
    'legend.fontsize': 12,
    'xtick.labelsize': 12,
    'ytick.labelsize': 12
})
# Matrix A from the given A^T
A = np.array([
    [1, 1, 1, 1],
    [-1, 1, -1, 1],
    [0, 2, 0, 2]
], dtype=float)

# Columns of A
a1 = A[:, 0]
a2 = A[:, 1]
a3 = A[:, 2]
a4 = A[:, 3]

# Gram-Schmidt
q1 = a1 / np.linalg.norm(a1)

v2 = a2 - np.dot(a2, q1) * q1
q2 = v2 / np.linalg.norm(v2)

print("A =\n", A)
print("\nColumns of A:")
print("a1 =", a1)
print("a2 =", a2)
print("a3 =", a3)
print("a4 =", a4)

print("\nOrthonormal basis from Gram-Schmidt:")
print("q1 =", q1)
print("q2 =", q2)

# 3D plot
fig = plt.figure(figsize=(10, 8))
ax = fig.add_subplot(111, projection='3d')

# Origin
origin = np.array([[0, 0, 0]])

# Plot original column vectors
vectors = [a1, a2, a3, a4]
labels = ['a1', 'a2', 'a3', 'a4']
colors = ['r', 'b', 'r', 'b']

for v, label, color in zip(vectors, labels, colors):
    ax.quiver(0, 0, 0, v[0], v[1], v[2], color=color, linewidth=2, arrow_length_ratio=0.08, label=label)

# Plot orthonormal vectors
ax.quiver(0, 0, 0, q1[0], q1[1], q1[2], color='k', linewidth=3, arrow_length_ratio=0.12, label='q1')
ax.quiver(0, 0, 0, q2[0], q2[1], q2[2], color='g', linewidth=3, arrow_length_ratio=0.12, label='q2')

# Labels and formatting
ax.set_xlabel('x')
ax.set_ylabel('y')
ax.set_zlabel('z')
ax.set_title('Columns of A and Orthonormal Basis from Gram-Schmidt')

ax.set_xlim([-0.5, 1.5])
ax.set_ylim([-1.5, 1.5])
ax.set_zlim([-0.5, 2.5])

ax.legend()
plt.show()
```
