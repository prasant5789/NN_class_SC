```python
import numpy as np
import matplotlib.pyplot as plt
from mpl_toolkits.mplot3d import Axes3D

# =========================
# Global Settings
# =========================
lims = (-3.5, 3.5)

TITLE_SIZE = 16
LABEL_SIZE = 14
TICK_SIZE = 12
FS = 14  # for manual z label


# Define Sets
# S1: line x1 = x2 = x3
t = np.linspace(-2, 2, 300)
x = t
y = t
z = t

# S2: plane x1 + x2 + x3 = 1
grid = np.linspace(-2, 2, 30)
X, Y = np.meshgrid(grid, grid)
Z_plane = 1 - X - Y

# Intersection point
s3 = np.array([1/3, 1/3, 1/3])

# Intersection  (S1 ∩ S2)
fig = plt.figure()
ax = fig.add_subplot(111, projection='3d')

# plot plane
ax.plot_surface(X, Y, Z_plane, alpha=0.35)

# plot line
ax.plot(x, y, z, linewidth=2)

# intersection point
ax.scatter(s3[0], s3[1], s3[2], color='k', s=60)

# annotate point
ax.text(
    s3[0]+0.1,
    s3[1]+0.1,
    s3[2]+0.1,
    r'$(\frac{1}{3},\frac{1}{3},\frac{1}{3})$',
    fontsize=14
)

ax.set_title("Intersection  (S1 ∩ S2)", fontsize=TITLE_SIZE)
ax.set_xlim(lims)
ax.set_ylim(lims)
ax.set_zlim(lims)

ax.set_xlabel("x1", fontsize=LABEL_SIZE, labelpad=10)
ax.set_ylabel("x2", fontsize=LABEL_SIZE, labelpad=10)

# Manual z-label like your example
fig.text(0.86, 0.55, r'$x_3$', fontsize=FS,
         rotation=90, ha='center', va='center')

ax.tick_params(axis='both', which='major', labelsize=TICK_SIZE)
ax.tick_params(axis='z', labelsize=TICK_SIZE)

plt.show()


# Minkowski Sum  (S1 ⊕ S2)

fig = plt.figure()
ax = fig.add_subplot(111, projection='3d')

# Multiple shifted planes
shifts = np.linspace(-2, 2, 8)
for shift in shifts:
    Z_shifted = Z_plane + shift
    ax.plot_surface(X, Y, Z_shifted, alpha=0.2)

# show S1 direction
ax.plot(x, y, z, linewidth=2)

ax.set_title("Minkowski Sum  (S1 ⊕ S2)", fontsize=TITLE_SIZE)
ax.set_xlim(lims)
ax.set_ylim(lims)
ax.set_zlim(lims)

ax.set_xlabel("x1", fontsize=LABEL_SIZE, labelpad=10)
ax.set_ylabel("x2", fontsize=LABEL_SIZE, labelpad=10)

# z-label
fig.text(0.86, 0.55, r'$x_3$', fontsize=FS,
         rotation=90, ha='center', va='center')

ax.tick_params(axis='both', which='major', labelsize=TICK_SIZE)
ax.tick_params(axis='z', labelsize=TICK_SIZE)

plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt
from mpl_toolkits.mplot3d import Axes3D   # registers 3D projection

FS = 18   # global font size used for labels/titles

def save_paraboloid():
    grid = np.linspace(-2, 2, 200)
    X, Y = np.meshgrid(grid, grid)
    Z = X**2 + Y**2

    fig = plt.figure(figsize=(7,5))
    ax = fig.add_subplot(111, projection='3d')
    ax.plot_surface(X, Y, Z, rstride=4, cstride=4, alpha=0.9)

    ax.set_title(r'$f(x)=x_1^2+x_2^2$', fontsize=FS, pad=20)
    ax.tick_params(axis='both', which='major', labelsize=16)
    ax.set_xlabel('$x_1$', fontsize=FS, labelpad=15)
    ax.set_ylabel('$x_2$', fontsize=FS, labelpad=15)

    
    fig.text(0.86, 0.55, r'$f(x_1,x_2)$', fontsize=FS, rotation=90, ha='center', va='center')

    ax.view_init(elev=30, azim=-60)
    plt.subplots_adjust(left=0.05, right=0.88, top=0.92, bottom=0.05)
    fig.savefig('paraboloid_fixed.png', dpi=200, pad_inches=0.18)
    plt.show()
    plt.close(fig)

def save_neg_xlnx():
    eps = 1e-3
    grid = np.linspace(eps, 3, 200)
    X, Y = np.meshgrid(grid, grid)
    Z = - X * np.log(X) - Y * np.log(Y)

    fig = plt.figure(figsize=(7,5))
    ax = fig.add_subplot(111, projection='3d')
    ax.plot_surface(X, Y, Z, rstride=4, cstride=4, alpha=0.9)

    ax.set_title(r'$f(x)=-x_1\ln x_1 - x_2\ln x_2$', fontsize=FS, pad=20)
    ax.tick_params(axis='both', which='major', labelsize=16)
    ax.set_xlabel('$x_1$', fontsize=FS, labelpad=15)
    ax.set_ylabel('$x_2$', fontsize=FS, labelpad=15)

    # stable z-label as figure text
    fig.text(0.86, 0.55, r'$f(x_1,x_2)$', fontsize=FS, rotation=90, ha='center', va='center')

    ax.view_init(elev=30, azim=-60)
    plt.subplots_adjust(left=0.05, right=0.88, top=0.92, bottom=0.05)
    fig.savefig('neg_xlnx_fixed.png', dpi=200, pad_inches=0.18)
    plt.show()
    plt.close(fig)

def save_l1_norm():
    grid = np.linspace(-2, 2, 200)
    X, Y = np.meshgrid(grid, grid)
    Z = np.abs(X) + np.abs(Y)

    fig = plt.figure(figsize=(7,5))
    ax = fig.add_subplot(111, projection='3d')
    ax.plot_surface(X, Y, Z, rstride=4, cstride=4, alpha=0.95)

    ax.set_title(r'$f(x)=|x_1|+|x_2|$', fontsize=FS, pad=20)
    ax.tick_params(axis='both', which='major', labelsize=16)
    ax.set_xlabel('$x_1$', fontsize=FS, labelpad=15)
    ax.set_ylabel('$x_2$', fontsize=FS, labelpad=15)

    # FIGURE-LEVEL Z LABEL (robust)
    fig.text(0.86, 0.55, r'$f(x_1,x_2)$', fontsize=FS, rotation=90, ha='center', va='center')

    # choose viewpoint you wanted
    ax.view_init(elev=30, azim=-45)

    plt.subplots_adjust(left=0.05, right=0.88, top=0.92, bottom=0.05)
    fig.savefig('l1_norm_fixed.png', dpi=200, pad_inches=0.18)
    plt.show()
    plt.close(fig)

if __name__ == '__main__':
    save_paraboloid()
    save_neg_xlnx()
    save_l1_norm()
    print("Saved: paraboloid_fixed.png, neg_xlnx_fixed.png, l1_norm_fixed.png")
```
