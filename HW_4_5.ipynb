```python
import numpy as np
import matplotlib.pyplot as plt

plt.rcParams.update({
    'font.size': 16,
    'axes.titlesize': 18,
    'axes.labelsize': 16,
    'legend.fontsize': 16,
    'xtick.labelsize': 16,
    'ytick.labelsize': 16
})

# -----------------------------
# Primal plot
# -----------------------------
x1 = np.linspace(0, 8, 500)
x2 = np.linspace(0, 6, 500)
X1, X2 = np.meshgrid(x1, x2)

# Primal feasible set
primal_feasible = (
    (-2*X1 - 3*X2 <= -4) &
    (-X1 + X2 <= -1) &
    (X1 >= 0) &
    (X2 >= 0)
)

plt.figure(figsize=(8, 6))
plt.contourf(X1, X2, primal_feasible, levels=[0.5, 1], alpha=0.35)

# Boundary lines
x2_line1 = (4 - 2*x1)/3          # 2x1 + 3x2 = 4
x2_line2 = x1 - 1                # x2 = x1 - 1

plt.plot(x1, x2_line1, label=r'$2x_1 + 3x_2 = 4$')
plt.plot(x1, x2_line2, label=r'$x_2 = x_1 - 1$')
plt.plot([2, 8], [0, 0], 'k--', label=r'feasible ray $(t,0),\, t\geq 2$')

plt.xlim(0, 8)
plt.ylim(0, 6)
plt.xlabel(r'$x_1$')
plt.ylabel(r'$x_2$')
plt.title('Primal Feasible Region')
plt.legend()
plt.grid(alpha=0.3)
plt.show()

# -----------------------------
# Dual plot
# -----------------------------
y1 = np.linspace(0, 4, 400)
y2 = np.linspace(0, 4, 400)
Y1, Y2 = np.meshgrid(y1, y2)

dual_feasible = (
    (-2*Y1 - Y2 >= 1) &
    (-3*Y1 + Y2 >= -1) &
    (Y1 >= 0) &
    (Y2 >= 0)
)

plt.figure(figsize=(8, 6))
plt.contourf(Y1, Y2, dual_feasible, levels=[0.5, 1], alpha=0.35)

# Constraint boundaries
y2_line1 = -2*y1 - 1             # -2y1 - y2 = 1
y2_line2 = 3*y1 - 1              # -3y1 + y2 = -1

plt.plot(y1, y2_line1, label=r'$-2y_1 - y_2 = 1$')
plt.plot(y1, y2_line2, label=r'$-3y_1 + y_2 = -1$')

plt.xlim(0, 4)
plt.ylim(0, 4)
plt.xlabel(r'$y_1$')
plt.ylabel(r'$y_2$')
plt.title('Dual Feasible Region (Empty)')
plt.legend()
plt.grid(alpha=0.3)
plt.show()

print("Conclusion:")
print("Primal is feasible and unbounded.")
print("Dual feasible region is empty (dual is infeasible).")
```
